```python
import re

text = "My email is student123@gmail.com and my phone is 9876543210."

# Search for email
email = re.search(r'\w+@\w+\.\w+', text)
print("Email:", email.group())

# Find all numbers
numbers = re.findall(r'\d+', text)
print("Numbers:", numbers)

# Match a word at the beginning
result = re.match(r'My', text)
print("Match:", result.group() if result else "No match")
```

```python
def accepts(string):
    state = 0

    for ch in string:
        if state == 0:
            state = 1 if ch == 'a' else 0
        elif state == 1:
            state = 2 if ch == 'b' else (1 if ch == 'a' else 0)
        elif state == 2:
            state = 1 if ch == 'a' else 0

    return state == 2


s = input("Enter a string: ")

if accepts(s):
    print("Accepted - string ends with 'ab'")
else:
    print("Rejected")
```

```python
def pluralize(word):
    if word.endswith(("s", "x", "z", "ch", "sh")):
        return word + "es"
    elif word.endswith("y") and word[-2] not in "aeiou":
        return word[:-1] + "ies"
    else:
        return word + "s"


words = ["cat", "box", "bus", "baby", "book"]

for word in words:
    print(word, "->", pluralize(word))
```

```python
import re

def pos_tag(word):
    if re.match(r'.*ing$', word):
        return "VBG"
    elif re.match(r'.*ed$', word):
        return "VBD"
    elif re.match(r'.*ly$', word):
        return "RB"
    elif re.match(r'.*s$', word):
        return "NNS"
    else:
        return "NN"


words = ["running", "played", "quickly", "books", "student"]

for word in words:
    print(word, "->", pos_tag(word))
```

```python
words = ["I", "am", "running", "fast"]

# Initial tags
tags = ["NN", "NN", "NN", "NN"]

# Transformation rules
for i, word in enumerate(words):
    if word.lower() in ["am", "is", "are"]:
        tags[i] = "VB"
    elif word.endswith("ing"):
        tags[i] = "VBG"
    elif word.lower() == "i":
        tags[i] = "PRP"

for word, tag in zip(words, tags):
    print(word, "->", tag)
```

```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V", "NP"]],
    "Det": [["the"]],
    "N": [["cat"], ["dog"]],
    "V": [["sees"]]
}

def parse(symbols, words):
    if not symbols:
        return len(words) == 0

    symbol = symbols[0]

    if symbol not in grammar:
        return words and words[0] == symbol and parse(symbols[1:], words[1:])

    for rule in grammar[symbol]:
        if parse(rule + symbols[1:], words):
            return True

    return False


sentence = "the cat sees the dog".split()

print("Accepted" if parse(["S"], sentence) else "Rejected")
```

```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V", "NP"]],
    "Det": [["the"]],
    "N": [["cat"], ["dog"]],
    "V": [["sees"]]
}

def earley(words):
    n = len(words)
    chart = [set() for _ in range(n + 1)]

    chart[0].add(("S", tuple(grammar["S"][0]), 0, 0))

    for i in range(n + 1):
        changed = True

        while changed:
            changed = False

            for lhs, rhs, dot, origin in list(chart[i]):

                if dot < len(rhs):
                    symbol = rhs[dot]

                    if symbol in grammar:
                        for rule in grammar[symbol]:
                            item = (symbol, tuple(rule), 0, i)
                            if item not in chart[i]:
                                chart[i].add(item)
                                changed = True

                    elif i < n and words[i] == symbol:
                        chart[i + 1].add((lhs, rhs, dot + 1, origin))

                else:
                    for item in list(chart[origin]):
                        l2, r2, d2, o2 = item
                        if d2 < len(r2) and r2[d2] == lhs:
                            new_item = (l2, r2, d2 + 1, o2)
                            if new_item not in chart[i]:
                                chart[i].add(new_item)
                                changed = True

    return ("S", tuple(grammar["S"][0]), 1, 0) in chart[n]


sentence = "the cat sees the dog".split()

print("Accepted" if earley(sentence) else "Rejected")
```

```python
def check_agreement(subject, verb):
    singular_subjects = ["he", "she", "it"]
    plural_subjects = ["they", "we"]

    singular_verbs = ["is", "has", "runs"]
    plural_verbs = ["are", "have", "run"]

    if subject in singular_subjects and verb in singular_verbs:
        return True

    if subject in plural_subjects and verb in plural_verbs:
        return True

    return False


subject = input("Enter subject: ").lower()
verb = input("Enter verb: ").lower()

if check_agreement(subject, verb):
    print("Agreement is correct")
else:
    print("Agreement is incorrect")
```

```python
import nltk

grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> 'the' 'cat' [0.5]
NP -> 'the' 'dog' [0.5]
VP -> 'runs' [0.6]
VP -> 'sleeps' [0.4]
""")

parser = nltk.ViterbiParser(grammar)

sentence = "the cat runs".split()

for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
```

```python
import re

def parse_fopc(expression):
    pattern = r'([A-Z][A-Za-z]*)\((.*?)\)'

    matches = re.findall(pattern, expression)

    if not matches:
        print("Invalid FOPC expression")
    else:
        for predicate, arguments in matches:
            print("Predicate:", predicate)
            print("Arguments:", arguments.split(","))


expression = "Likes(John, Mary)"

parse_fopc(expression)
```

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'student' | 'book'
V -> 'reads'
""")

parser = nltk.ChartParser(grammar)

sentence = "the student reads the book".split()

for tree in parser.parse(sentence):
    print(tree)

    noun_phrases = tree.subtrees(
        filter=lambda t: t.label() == "NP"
    )

    print("Noun Phrases:")
    for np in noun_phrases:
        print(" ".join(np.leaves()))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text = [
    "Machine learning is a branch of artificial intelligence.",
    "Artificial intelligence allows computers to learn from data.",
    "Machine learning algorithms are used in many applications."
]

vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform(text)

scores = []

for i in range(len(text) - 1):
    score = cosine_similarity(
        vectors[i:i+1], vectors[i+1:i+2]
    )[0][0]
    scores.append(score)

print("Sentence similarity scores:")

for i, score in enumerate(scores):
    print(f"Sentence {i+1} -> {i+2}: {score:.3f}")

print("Coherence score:", sum(scores) / len(scores))
```

```python
def dialog_act(sentence):
    sentence = sentence.lower()

    if sentence.endswith("?"):
        return "QUESTION"

    elif any(word in sentence for word in ["hello", "hi", "hey"]):
        return "GREETING"

    elif any(word in sentence for word in ["thanks", "thank you"]):
        return "THANKING"

    elif any(word in sentence for word in ["please", "could you", "can you"]):
        return "REQUEST"

    elif any(word in sentence for word in ["bye", "goodbye"]):
        return "GOODBYE"

    else:
        return "STATEMENT"


dialogue = [
    "Hello",
    "Can you help me?",
    "Please send the file",
    "Thank you",
    "Goodbye"
]

for sentence in dialogue:
    print(sentence, "->", dialog_act(sentence))
```
